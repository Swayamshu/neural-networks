```python
b = {}

for w in words:
    chs = ["<S>"] + list(w) + ["<E>"] # this will prompt the model to think about the starting and ending of words
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
```

```python
import torch
```

```python
N = torch.zeros((28, 28), dtype=torch.int32)
```

```python
chars = sorted(list(set(''.join(words))))
stoi = {s:i for i,s in enumerate(chars)}
stoi['<S>'] = 26
stoi['<E>'] = 27
```

```python
for w in words:
    chs = ["<S>"] + list(w) + ["<E>"]
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1][ix2] += 1
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.imshow(N)
```
